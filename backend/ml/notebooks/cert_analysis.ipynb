```python
import pandas as pd
from collections import Counter

# ---------------------------------------------------------
# CERT R4.2 Behavioral Profile
# ---------------------------------------------------------

activity_files = {
    "logon": "logon.csv",
    "email": "email.csv",
    "file": "file.csv",
    "http": "http.csv",
    "device": "device.csv",
}

for name, filename in activity_files.items():

    file_path = DATASET_PATH / filename

    print("\n" + "=" * 70)
    print(f"{name.upper()} DATASET")
    print("=" * 70)

    unique_users = set()
    unique_pcs = set()
    activity_counts = Counter()

    min_date = None
    max_date = None

    for chunk in pd.read_csv(
        file_path,
        chunksize=100_000,
        usecols=["date", "user", "pc"] +
                (["activity"] if name in ["logon", "device"] else [])
    ):

        # -----------------------------
        # Users
        # -----------------------------
        unique_users.update(
            chunk["user"].dropna().astype(str).unique()
        )

        # -----------------------------
        # PCs
        # -----------------------------
        unique_pcs.update(
            chunk["pc"].dropna().astype(str).unique()
        )

        # -----------------------------
        # Activity types
        # -----------------------------
        if "activity" in chunk.columns:
            activity_counts.update(
                chunk["activity"]
                .dropna()
                .astype(str)
                .value_counts()
                .to_dict()
            )

        # -----------------------------
        # Dates
        # -----------------------------
        dates = pd.to_datetime(
            chunk["date"],
            errors="coerce"
        )

        chunk_min = dates.min()
        chunk_max = dates.max()

        if pd.notna(chunk_min):
            if min_date is None or chunk_min < min_date:
                min_date = chunk_min

        if pd.notna(chunk_max):
            if max_date is None or chunk_max > max_date:
                max_date = chunk_max

    print(f"Unique users : {len(unique_users):,}")
    print(f"Unique PCs   : {len(unique_pcs):,}")
    print(f"Start date   : {min_date}")
    print(f"End date     : {max_date}")

    if activity_counts:
        print("\nActivity types:")
        for activity, count in activity_counts.items():
            print(f"  {activity:15} {count:,}")
```

```python
import pandas as pd

logon_features = {}

for chunk in pd.read_csv(
    DATASET_PATH / "logon.csv",
    chunksize=100_000,
    usecols=["date", "user", "pc", "activity"],
    parse_dates=["date"]
):
    # Remove records without a user
    chunk = chunk.dropna(subset=["user"])

    # Convert user to string consistently
    chunk["user"] = chunk["user"].astype(str)

    # Identify logon/logoff
    is_logon = chunk["activity"].eq("Logon")
    is_logoff = chunk["activity"].eq("Logoff")

    # After-hours:
    # before 08:00 OR at/after 18:00
    hour = chunk["date"].dt.hour
    after_hours = (hour < 8) | (hour >= 18)

    # Aggregate this chunk by user
    chunk_features = (
        chunk.assign(
            logon_count=is_logon.astype(int),
            logoff_count=is_logoff.astype(int),
            after_hours_logon=(
                is_logon & after_hours
            ).astype(int),
            after_hours_logoff=(
                is_logoff & after_hours
            ).astype(int)
        )
        .groupby("user")
        .agg(
            logon_count=("logon_count", "sum"),
            logoff_count=("logoff_count", "sum"),
            after_hours_logon_count=(
                "after_hours_logon",
                "sum"
            ),
            after_hours_logoff_count=(
                "after_hours_logoff",
                "sum"
            ),
            unique_pc_count=("pc", "nunique")
        )
    )

    # Merge this chunk into our running result
    for user, row in chunk_features.iterrows():

        if user not in logon_features:
            logon_features[user] = row.to_dict()
        else:
            logon_features[user]["logon_count"] += row["logon_count"]
            logon_features[user]["logoff_count"] += row["logoff_count"]
            logon_features[user]["after_hours_logon_count"] += (
                row["after_hours_logon_count"]
            )
            logon_features[user]["after_hours_logoff_count"] += (
                row["after_hours_logoff_count"]
            )

            # unique PC needs a proper set-based calculation,
            # so we handle it separately below.
```

```python
# ============================================================
# EMAIL FEATURE ENGINEERING
# ============================================================

from collections import defaultdict

email_counts = defaultdict(int)
email_attachment_count = defaultdict(int)
email_total_attachments = defaultdict(int)
email_total_size = defaultdict(int)
email_recipients = defaultdict(set)

email_file = DATASET_PATH / "email.csv"

for chunk in pd.read_csv(
    email_file,
    chunksize=100_000,
    usecols=[
        "user",
        "to",
        "cc",
        "bcc",
        "size",
        "attachments"
    ]
):
    # Remove rows without a user
    chunk = chunk.dropna(subset=["user"])

    chunk["user"] = chunk["user"].astype(str)

    # --------------------------------------------------------
    # Email count
    # --------------------------------------------------------
    counts = chunk["user"].value_counts()

    for user, count in counts.items():
        email_counts[user] += int(count)

    # --------------------------------------------------------
    # Attachments
    # --------------------------------------------------------
    chunk["attachments"] = pd.to_numeric(
        chunk["attachments"],
        errors="coerce"
    ).fillna(0)

    attachment_rows = chunk["attachments"] > 0

    attachment_counts = (
        chunk.loc[attachment_rows, "user"]
        .value_counts()
    )

    for user, count in attachment_counts.items():
        email_attachment_count[user] += int(count)

    attachment_totals = (
        chunk.groupby("user")["attachments"]
        .sum()
    )

    for user, total in attachment_totals.items():
        email_total_attachments[user] += int(total)

    # --------------------------------------------------------
    # Email size
    # --------------------------------------------------------
    chunk["size"] = pd.to_numeric(
        chunk["size"],
        errors="coerce"
    ).fillna(0)

    size_totals = (
        chunk.groupby("user")["size"]
        .sum()
    )

    for user, total in size_totals.items():
        email_total_size[user] += int(total)

    # --------------------------------------------------------
    # Recipients
    # --------------------------------------------------------
    for column in ["to", "cc", "bcc"]:

        recipients = chunk[["user", column]].dropna()

        for user, recipient in zip(
            recipients["user"],
            recipients[column]
        ):
            email_recipients[user].add(
                str(recipient)
            )

print("Email processing completed.")
print("Users found:", len(email_counts))
```

```python
email_features_df = pd.DataFrame({
    "user": list(email_counts.keys()),
    "email_count": [
        email_counts[user]
        for user in email_counts
    ],
    "emails_with_attachments": [
        email_attachment_count[user]
        for user in email_counts
    ],
    "total_attachments": [
        email_total_attachments[user]
        for user in email_counts
    ],
    "total_email_size": [
        email_total_size[user]
        for user in email_counts
    ],
    "unique_recipients": [
        len(email_recipients[user])
        for user in email_counts
    ]
})

display(email_features_df.head(10))
print("Shape:", email_features_df.shape)
```

```python
# ============================================================
# FILE FEATURE ENGINEERING
# ============================================================

from collections import defaultdict

file_activity_count = defaultdict(int)
file_names = defaultdict(set)
file_pcs = defaultdict(set)

file_path = DATASET_PATH / "file.csv"

for chunk in pd.read_csv(
    file_path,
    chunksize=100_000,
    usecols=["user", "pc", "filename"]
):
    # Remove records without a user
    chunk = chunk.dropna(subset=["user"])

    chunk["user"] = chunk["user"].astype(str)

    # --------------------------------------------------------
    # Total file activity
    # --------------------------------------------------------
    counts = chunk["user"].value_counts()

    for user, count in counts.items():
        file_activity_count[user] += int(count)

    # --------------------------------------------------------
    # Unique filenames
    # --------------------------------------------------------
    filename_rows = chunk.dropna(subset=["filename"])

    for user, filename in zip(
        filename_rows["user"],
        filename_rows["filename"]
    ):
        file_names[user].add(str(filename))

    # --------------------------------------------------------
    # Unique PCs
    # --------------------------------------------------------
    pc_rows = chunk.dropna(subset=["pc"])

    for user, pc in zip(
        pc_rows["user"],
        pc_rows["pc"]
    ):
        file_pcs[user].add(str(pc))

print("File processing completed.")
print("Users found:", len(file_activity_count))
```

```python
# ============================================================
# HTTP FEATURE ENGINEERING
# ============================================================

from collections import defaultdict
from urllib.parse import urlparse

http_request_count = defaultdict(int)
http_pcs = defaultdict(set)
http_domains = defaultdict(set)

http_file = DATASET_PATH / "http.csv"

for chunk_number, chunk in enumerate(
    pd.read_csv(
        http_file,
        chunksize=100_000,
        usecols=["user", "pc", "url"]
    ),
    start=1
):

    # Remove records without user
    chunk = chunk.dropna(subset=["user"])
    chunk["user"] = chunk["user"].astype(str)

    # --------------------------------------------------------
    # HTTP request count
    # --------------------------------------------------------
    counts = chunk["user"].value_counts()

    for user, count in counts.items():
        http_request_count[user] += int(count)

    # --------------------------------------------------------
    # Unique PCs
    # --------------------------------------------------------
    pc_rows = chunk.dropna(subset=["pc"])

    for user, pc in zip(
        pc_rows["user"],
        pc_rows["pc"]
    ):
        http_pcs[user].add(str(pc))

    # --------------------------------------------------------
    # Extract domains instead of storing complete URLs
    # --------------------------------------------------------
    url_rows = chunk.dropna(subset=["url"])

    for user, url in zip(
        url_rows["user"],
        url_rows["url"]
    ):
        try:
            url_string = str(url)

            # Add scheme if missing
            if not url_string.startswith(
                ("http://", "https://")
            ):
                url_string = "http://" + url_string

            domain = urlparse(
                url_string
            ).netloc.lower()

            if domain:
                http_domains[user].add(domain)

        except Exception:
            continue

    if chunk_number % 25 == 0:
        print(
            f"Processed approximately "
            f"{chunk_number * 100_000:,} rows..."
        )

print("\nHTTP processing completed.")
print("Users found:", len(http_request_count))
```

```python
# ============================================================
# DEVICE FEATURE ENGINEERING
# ============================================================

from collections import defaultdict

device_activity_count = defaultdict(int)
device_connect_count = defaultdict(int)
device_disconnect_count = defaultdict(int)
device_pcs = defaultdict(set)

device_file = DATASET_PATH / "device.csv"

for chunk in pd.read_csv(
    device_file,
    chunksize=100_000,
    usecols=["user", "pc", "activity"]
):

    chunk = chunk.dropna(subset=["user"])
    chunk["user"] = chunk["user"].astype(str)

    # --------------------------------------------------------
    # Total device activity
    # --------------------------------------------------------
    counts = chunk["user"].value_counts()

    for user, count in counts.items():
        device_activity_count[user] += int(count)

    # --------------------------------------------------------
    # Connect / Disconnect
    # --------------------------------------------------------
    connect_counts = (
        chunk.loc[
            chunk["activity"].eq("Connect"),
            "user"
        ].value_counts()
    )

    disconnect_counts = (
        chunk.loc[
            chunk["activity"].eq("Disconnect"),
            "user"
        ].value_counts()
    )

    for user, count in connect_counts.items():
        device_connect_count[user] += int(count)

    for user, count in disconnect_counts.items():
        device_disconnect_count[user] += int(count)

    # --------------------------------------------------------
    # Unique PCs
    # --------------------------------------------------------
    pc_rows = chunk.dropna(subset=["pc"])

    for user, pc in zip(
        pc_rows["user"],
        pc_rows["pc"]
    ):
        device_pcs[user].add(str(pc))

print("Device processing completed.")
print("Users found:", len(device_activity_count))
```

```python
# ============================================================
# ML FEATURE PREPARATION
# ============================================================

print("Master dataset shape:", master_features_df.shape)

print("\nData types:")
print(master_features_df.dtypes)

print("\nMissing values:")
print(master_features_df.isna().sum())

print("\nNumber of numeric columns:",
      master_features_df.select_dtypes(
          include="number"
      ).shape[1])
```

```python
# ============================================================
# ML FEATURE MATRIX
# ============================================================

# Columns that identify the employee/user.
ID_COLUMNS = [
    "employee_name",
    "user"
]

# Select only numeric behavioral/personality features.
X = master_features_df.drop(
    columns=ID_COLUMNS
).copy()

print("Feature matrix shape:", X.shape)

print("\nFeature columns:")
for column in X.columns:
    print("-", column)
```

```python
# Check missing values
print("Total missing values:", X.isna().sum().sum())

# Check infinite values
print(
    "Total infinite values:",
    X.isin([float("inf"), float("-inf")]).sum().sum()
)

# Check constant features
constant_features = [
    column
    for column in X.columns
    if X[column].nunique() <= 1
]

print("\nConstant features:")
print(constant_features)
```

```python
# ============================================================
# REMOVE CONSTANT FEATURES FROM ML INPUT
# ============================================================

constant_features = [
    column
    for column in X.columns
    if X[column].nunique() <= 1
]

print("Constant features:", constant_features)

X_model = X.drop(
    columns=constant_features
).copy()

print("\nOriginal feature count:", X.shape[1])
print("ML feature count:", X_model.shape[1])
print("Removed:", constant_features)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_model)

print("Scaled feature matrix shape:", X_scaled.shape)
```

```python
# ============================================================
# ISOLATION FOREST - INSIDER THREAT ANOMALY DETECTION
# ============================================================

from sklearn.ensemble import IsolationForest

model = IsolationForest(
    n_estimators=300,
    contamination=0.05,
    random_state=42,
    n_jobs=-1
)

model.fit(X_scaled)

print("Isolation Forest training completed.")
print("Number of trees:", model.n_estimators)
print("Contamination:", model.contamination)
```

```python
# ============================================================
# GENERATE ANOMALY SCORES
# ============================================================

raw_scores = model.decision_function(X_scaled)

predictions = model.predict(X_scaled)

print("Anomaly scores generated.")

print("\nPrediction counts:")
print(pd.Series(predictions).value_counts())
```

```python
print("Minimum score:", raw_scores.min())
print("Maximum score:", raw_scores.max())
print("Mean score:", raw_scores.mean())
```

```python
# ============================================================
# CONVERT ANOMALY SCORE → RISK SCORE
# ============================================================

anomaly_strength = -raw_scores

min_score = anomaly_strength.min()
max_score = anomaly_strength.max()

if max_score == min_score:
    risk_scores = pd.Series(
        0.0,
        index=master_features_df.index
    )
else:
    risk_scores = pd.Series(
        (
            (anomaly_strength - min_score)
            / (max_score - min_score)
            * 100
        ),
        index=master_features_df.index
    )

risk_scores = risk_scores.round(2)

print("Risk score generated.")

print("\nRisk score statistics:")
print(risk_scores.describe())
```

```python
# ============================================================
# RISK LEVEL
# ============================================================

def get_risk_level(score):
    if score < 25:
        return "Low"
    elif score < 50:
        return "Medium"
    elif score < 75:
        return "High"
    else:
        return "Critical"


risk_levels = risk_scores.apply(get_risk_level)

print(risk_levels.value_counts())
```

```python
# ============================================================
# FINAL RISK RESULTS
# ============================================================

risk_results_df = master_features_df[
    ["employee_name", "user"]
].copy()

risk_results_df["anomaly_score"] = raw_scores
risk_results_df["risk_score"] = risk_scores
risk_results_df["risk_level"] = risk_levels
risk_results_df["is_anomaly"] = (
    predictions == -1
)

display(
    risk_results_df
    .sort_values(
        "risk_score",
        ascending=False
    )
    .head(20)
)
```
